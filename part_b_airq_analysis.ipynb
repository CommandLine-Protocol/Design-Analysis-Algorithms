# Part B

```python
MONTHS = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
] #in months

PM25_DATA = [23, 31, 34, 42, 40, 42, 27, 23, 27, 26, 23, 31]       # in μg/m³
TEMP_DATA = [40, 43, 48, 53, 64, 68, 75, 48, 48, 60, 53, 40]       # in °F
HUMIDITY_DATA = [26, 25, 29, 38, 42, 38, 34, 42, 38, 32, 31, 26]   # in %
```

```python
# Combined 2D Array: Rows = 12 months, Cols = [PM2.5, Temperature, Humidity]
WEATHER_MATRIX = [
    [23, 40, 26],  # January
    [31, 43, 25],  # February
    [34, 48, 29],  # March
    [42, 53, 38],  # April
    [40, 64, 42],  # May
    [42, 68, 38],  # June
    [27, 75, 34],  # July
    [23, 48, 42],  # August
    [27, 48, 38],  # September
    [26, 60, 32],  # October
    [23, 53, 31],  # November
    [31, 40, 26],  # December
]
```

### Number 1: month with the highest PM2.5 concentration

```python
highest_conc = max(PM25_DATA)

highest_months = [
    MONTHS[i]
    for i, value in enumerate(PM25_DATA)
    if value == highest_conc
]

print("Months with the highest PM2.5 concentration:", highest_months)
```

### Number 2: average PM2.5 concentration

```python
total_conc = 0
for row in WEATHER_MATRIX:
    total_conc += row[0]  # Add PM2.5 concentration for each month
print(f"Total PM2.5 concentration: {total_conc} μg/m³")
```

```python
average_conc = total_conc / len(WEATHER_MATRIX)
print(f"Average PM2.5 concentration: {average_conc:.2f} μg/m³")
```

### Number 3: month with the highest humidity

```python
highest_humidity = max(HUMIDITY_DATA)

highest_months = [
    MONTHS[i]
    for i, value in enumerate(HUMIDITY_DATA)
    if value == highest_humidity
]

print("Months with the highest Humidity:", highest_months)
```

### Number 4: average temperature for the year

```python
average_temp = sum(TEMP_DATA) / len(TEMP_DATA)
print(f"Average Temperature: {average_temp:.2f} °F")
```

### Number 5: month with the lowest PM2.5 concentration

```python
lowest_conc = min(PM25_DATA)

lowest_months = [
    MONTHS[i]
    for i, value in enumerate(PM25_DATA)
    if value == lowest_conc
]

print("Months with the lowest PM2.5 concentration:", lowest_months)
```

### Number 6: months where the PM2.5 concentration was above the annual average

```python
print("Months with PM2.5 concentration above average:")
for month_index, row in enumerate(WEATHER_MATRIX):
    if row[0] > average_conc:
        print(f"{MONTHS[month_index]}")
```

### Number 7

### I. store all the monthly PM2.5 measurements in the Dictionary

```python
# I. Store the monthly measurements in a dictionary

PM25_MONTHLY = dict(zip(MONTHS, PM25_DATA))

print(PM25_MONTHLY)
```

### II. Prompts the user to enter a month and displays the corresponding PM2.5 concentration.

```python
# II. Ask the user for a month

month = input("Enter a month: ").strip().title()

if month in PM25_MONTHLY:
    print("Month:", month)
    print("PM2.5 concentration:", PM25_MONTHLY[month], "µg/m³")
else:
    print("Month not found.")
```

### III. Finds and displays the month with the highest PM2.5 concentration

```python
# III. Find the highest concentration and its month(s)

highest_concentration = max(PM25_MONTHLY.values())

highest_months = []

for month in PM25_MONTHLY:
    if PM25_MONTHLY[month] == highest_concentration:
        highest_months.append(month)

print("Month(s) with the highest PM2.5:", highest_months)
print("Highest concentration:", highest_concentration, "µg/m³")
```

### IV. Calculates and displays the average PM2.5 concentration for all months.

```python
# IV. Calculate the average

total = sum(PM25_MONTHLY.values())
average = total / len(PM25_MONTHLY)

print("Average PM2.5 concentration:", average, "µg/m³")
```

### Number 8

### I. unique values recorded for each of the three environmental variables.

```python
PM25_DATA_SET = set(PM25_DATA)
TEMP_DATA_SET = set(TEMP_DATA)
HUMIDITY_DATA_SET = set(HUMIDITY_DATA)

print(f"Unique PM2.5 concentrations: {PM25_DATA_SET}")
print(f"Unique Temperatures: {TEMP_DATA_SET}")
print(f"Unique Humidity levels: {HUMIDITY_DATA_SET}")
```

### II. number of unique values recorded for PM2.5, temperature, and humidity

```python
num_unique_pm25 = len(PM25_DATA_SET)
num_unique_temp = len(TEMP_DATA_SET)
num_unique_humidity = len(HUMIDITY_DATA_SET)

print(f"Number of unique PM2.5 concentrations: {num_unique_pm25}")
print(f"Number of unique Temperatures: {num_unique_temp}")
print(f"Number of unique Humidity levels: {num_unique_humidity}")
```

### III. which of the three variables has the highest number of repeated values during the year

```python
unique_stats = [
    ("humidity", num_unique_humidity),
    ("PM2.5 concentration", num_unique_pm25),
    ("temperature", num_unique_temp)
]

least = unique_stats[0]

for variable, count in unique_stats:
    if count < least[1]:
        least = (variable, count)

print(f"The variable with the least number of unique values is {least[0]} with {least[1]} unique values.")
```
