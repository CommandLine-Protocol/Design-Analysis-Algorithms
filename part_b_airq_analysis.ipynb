```python
MONTHS = [
    "January", "February", "March", "April", "May", "June",
    "July", "August", "September", "October", "November", "December"
] #in months

PM25_DATA = [23, 31, 34, 42, 40, 42, 27, 23, 27, 26, 23, 31]       # in μg/m³
TEMP_DATA = [40, 43, 48, 53, 64, 68, 75, 48, 48, 60, 53, 40]       # in °F
HUMIDITY_DATA = [26, 25, 29, 38, 42, 38, 34, 42, 38, 32, 31, 26]   # in %
```

```python
# Combined 2D Array: Rows = 12 months, Cols = [PM2.5, Temperature, Humidity]
WEATHER_MATRIX = [
    [23, 40, 26],  # January
    [31, 43, 25],  # February
    [34, 48, 29],  # March
    [42, 53, 38],  # April
    [40, 64, 42],  # May
    [42, 68, 38],  # June
    [27, 75, 34],  # July
    [23, 48, 42],  # August
    [27, 48, 38],  # September
    [26, 60, 32],  # October
    [23, 53, 31],  # November
    [31, 40, 26],  # December
]
```

```python
total_conc = 0
for row in WEATHER_MATRIX:
    total_conc += row[0]  # Add PM2.5 concentration for each month
print(f"Total PM2.5 concentration: {total_conc} μg/m³")
```

```python
average_conc = total_conc / len(WEATHER_MATRIX)
print(f"Average PM2.5 concentration: {average_conc:.2f} μg/m³")
```

```python
highest_humidity = max(row[2] for row in WEATHER_MATRIX)
for month_index, row in enumerate(WEATHER_MATRIX):
    if row[2] == highest_humidity:
        print(f"Month with highest humidity: {MONTHS[month_index]} ({highest_humidity}%)")
        break
```

```python
average_temp = sum(TEMP_DATA) / len(TEMP_DATA)
print(f"Average Temperature: {average_temp:.2f} °F")
```

```python
lowest_conc = min(row[0] for row in WEATHER_MATRIX)
for month_index, row in enumerate(WEATHER_MATRIX):
    if row[0] == lowest_conc:
        print(f"Month with lowest PM2.5 concentration: {MONTHS[month_index]} ({lowest_conc} μg/m³)")
        break
```

```python
print("Months with PM2.5 concentration above average:")
for month_index, row in enumerate(WEATHER_MATRIX):
    if row[0] > average_conc:
        print(f"{MONTHS[month_index]}")
```

```python
PM25_DATA_SET = set(PM25_DATA)
TEMP_DATA_SET = set(TEMP_DATA)
HUMIDITY_DATA_SET = set(HUMIDITY_DATA)

print(f"Unique PM2.5 concentrations: {PM25_DATA_SET}")
print(f"Unique Temperatures: {TEMP_DATA_SET}")
print(f"Unique Humidity levels: {HUMIDITY_DATA_SET}")
```

```python
unique_stats = [num_unique_humidity, num_unique_pm25, num_unique_temp]
min = unique_stats[0]
for i in unique_stats:
    if i < min and i == 1:
        min = i
        print(f"The variable with the least number of unique values is PM2.5 concentration with {min} unique values.")
        break
    elif i < min and i == 2:
        min = i
        print(f"The variable with the least number of unique values is temperature with {min} unique values.")
        break
```

```python
# III. Find the month with the highest PM2.5 concentration
highest_month = max(PM2_5, key=PM2_5.get)

print("Month with the highest PM2.5 concentration:", highest_month)
print("Highest concentration:", PM2_5[highest_month], "μg/m³")
```
